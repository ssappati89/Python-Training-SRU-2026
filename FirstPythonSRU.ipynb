````python
# Get three numbers from user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

# Check if num1 is prime
if num1 <= 1:
    prime1 = False
else:
    prime1 = True
    i = 2
    while i < num1:
        if num1 % i == 0:
            prime1 = False
            break
        i = i + 1

# Check if num2 is prime
if num2 <= 1:
    prime2 = False
else:
    prime2 = True
    i = 2
    while i < num2:
        if num2 % i == 0:
            prime2 = False
            break
        i = i + 1

# Check if num3 is prime
if num3 <= 1:
    prime3 = False
else:
    prime3 = True
    i = 2
    while i < num3:
        if num3 % i == 0:
            prime3 = False
            break
        i = i + 1

# Find lowest prime using simple if-else
lowest = None

if prime1:
    lowest = num1

if prime2:
    if lowest is None:
        lowest = num2
    else:
        if num2 < lowest:
            lowest = num2

if prime3:
    if lowest is None:
        lowest = num3
    else:
        if num3 < lowest:
            lowest = num3

# Display result
if lowest is not None:
    print(f"Lowest prime number: {lowest}")
else:
    print("No prime numbers found.")
````

### Centering Text in Python

To center-align text output in Python, you can use the string method `.center(width, fillchar)`.

*   `width`: The total width of the string that you want to achieve.
*   `fillchar`: (Optional) The character used to fill the padding around the centered string. If not specified, spaces are used by default.

````python
# Demonstrate center-aligned output
# We'll choose a total width, for example, 10 characters.
max_width = 11

for i in range(7):
  # Create the string of 'A's
  a_string = "A" * (i + 1)
  # Center the string within the specified width, using '_' as the fill character
  print(a_string.center(max_width, '_'))

print("\n--- Right Aligned Example ---")
for i in range(5):
  a_string = "A" * (i + 1)
  print(a_string.rjust(max_width))

print("\n--- Left Aligned Example (Original) ---")
for i in range(5):
  a_string = "A" * (i + 1)
  print(a_string.ljust(max_width)) # Or simply print(a_string)
````
