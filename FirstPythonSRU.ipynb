```python
# Get three numbers from user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

# Check if num1 is prime
if num1 <= 1:
    prime1 = False
else:
    prime1 = True
    i = 2
    while i < num1:
        if num1 % i == 0:
            prime1 = False
            break
        i = i + 1

# Check if num2 is prime
if num2 <= 1:
    prime2 = False
else:
    prime2 = True
    i = 2
    while i < num2:
        if num2 % i == 0:
            prime2 = False
            break
        i = i + 1

# Check if num3 is prime
if num3 <= 1:
    prime3 = False
else:
    prime3 = True
    i = 2
    while i < num3:
        if num3 % i == 0:
            prime3 = False
            break
        i = i + 1

# Find lowest prime using simple if-else
lowest = None

if prime1:
    lowest = num1

if prime2:
    if lowest is None:
        lowest = num2
    else:
        if num2 < lowest:
            lowest = num2

if prime3:
    if lowest is None:
        lowest = num3
    else:
        if num3 < lowest:
            lowest = num3

# Display result
if lowest is not None:
    print(f"Lowest prime number: {lowest}")
else:
    print("No prime numbers found.")
```
